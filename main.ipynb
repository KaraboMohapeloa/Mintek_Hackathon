# Explainable Probabilistic Decision Support for EAF Process Optimisation

## Proof-of-Concept:RAPPG


This notebook implements a simple end-to-end proof-of-concept for
probabilistic decision support in Electric Arc Furnace (EAF) steelmaking.

The POC investigates whether historical EAF process data can be used to:

1. Represent the current state of a heat.
2. Predict the distribution of possible tap temperatures.
3. Quantify the probability of achieving a target tap-temperature range.
4. Evaluate simple alternative operating scenarios.
5. Compare the scenarios using probability of success, energy and time.
6. Produce an interpretable decision-support summary.

The system is intended to support the furnace operator rather than
automatically control the EAF.

### Core concept

Current EAF State
        ↓
Bayesian Probabilistic Model
        ↓
Posterior Predictive Distribution
        ↓
Candidate Action Scenarios
        ↓
Probability of Target Achievement
        ↓
Energy / Time Considerations
        ↓
Decision-Support Output


## 1. Problem Statement

Electric Arc Furnace operation requires decisions to be made under
uncertain and changing process conditions.

During the final stages of a heat, the operator must determine whether
to maintain the current operation, modify process inputs, or proceed
toward tapping in order to achieve the desired tap-temperature range.

The challenge is that the outcome associated with a potential operating
scenario is uncertain and depends on the current state of the heat.

This POC therefore investigates a Bayesian probabilistic decision-support
approach that uses historical EAF process data to estimate the distribution
of possible tap temperatures from the current heat state.

Candidate operating scenarios are then evaluated using their predicted
temperature distributions and the probability of achieving the desired
temperature range.

The system is designed as operator decision support and not as an
autonomous furnace controller.
